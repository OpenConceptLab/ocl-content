```python
import json
import csv
```

```python
# Define the path to your JSON file
json_file_path  = "C:/Users/jamlung/Downloads/Regenstrief_LOINC_v2.71.21AA.2022-02-17_170852/export.json"
csv_file_path = 'concept_ids.csv'

try:
    # --- 1. Read the JSON file ---
    with open(json_file_path, 'r') as file:
        data = json.load(file)

    # --- 2. Extract the list of concepts ---
    concepts_list = data['concepts']

    # --- 3. Write the IDs to a CSV file ---
    with open(csv_file_path, 'w', newline='') as csvfile:
        # Create a CSV writer object
        csv_writer = csv.writer(csvfile)

        # Write the header row
        csv_writer.writerow(['id'])

        # Loop through each concept and write its ID to a new row
        for concept in concepts_list:
            csv_writer.writerow([concept['id']])

    print(f"✅ Success! Wrote {len(concepts_list)} IDs to '{csv_file_path}'")

except FileNotFoundError:
    print(f"❌ Error: The file '{json_file_path}' was not found.")
except KeyError:
    print("❌ Error: The key 'concepts' was not found in the JSON file.")
except Exception as e:
    print(f"❌ An unexpected error occurred: {e}")
```
