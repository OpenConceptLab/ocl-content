```python
import json
import csv
```

```python
# Define the path to your JSON file
json_file_path  = "C:/Users/jamlung/Downloads/Regenstrief_LOINC_v2.71.21AA.2022-02-17_170852/export.json"
csv_file_path = 'concept_ids-271.csv'

try:
    # --- 1. Read the JSON file ---
    with open(json_file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)

    # --- 2. Extract the list of concepts ---
    concepts_list = data.get('concepts', []) # Use .get() for safer access

    # --- 3. Write the ID and retired status to a CSV file ---
    with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
        # Create a CSV writer object
        csv_writer = csv.writer(csvfile)

        # Write the new header row
        csv_writer.writerow(['id', 'retired'])

        # Loop through each concept and write its attributes to a new row
        for concept in concepts_list:
            # Safely get the 'retired' attribute, defaulting to False if not present
            retired_status = concept.get('retired', False)
            csv_writer.writerow([concept['id'], retired_status])

    print(f"✅ Success! Wrote {len(concepts_list)} concepts (ID and retired status) to '{csv_file_path}'")

except FileNotFoundError:
    print(f"❌ Error: The file '{json_file_path}' was not found.")
except KeyError:
    print("❌ Error: A required key was not found in the JSON file. Ensure 'concepts' and 'id' keys exist.")
except Exception as e:
    print(f"❌ An unexpected error occurred: {e}")
```

```python
# Mapping ids

# Define the path to your JSON and output CSV files
json_file_path  = "C:/Users/jamlung/Downloads/Regenstrief_LOINC_v2.71.21AA.2022-02-17_170852/export.json"
csv_file_path = 'mapping_attributes-2-71.csv'


try:
    # --- 1. Read the JSON file ---
    with open(json_file_path, 'r', encoding='utf-8') as file:
        data = json.load(file)

    # --- 2. Extract the list of mappings ---
    mappings_list = data.get('mappings', []) # Target 'mappings' instead of 'concepts'

    # --- 3. Write the attributes to a CSV file ---
    with open(csv_file_path, 'w', newline='', encoding='utf-8') as csvfile:
        # Create a CSV writer object
        csv_writer = csv.writer(csvfile)

        # Define and write the header row for the desired attributes
        header = ['id', 'retired', 'from_concept_url', 'to_concept_url','map_type']
        csv_writer.writerow(header)

        # Loop through each mapping and write its attributes to a new row
        for mapping in mappings_list:
            # Use .get() to safely extract each value, providing a default
            row_data = [
                mapping.get('id'),
                mapping.get('retired', False),
                mapping.get('from_concept_url'),
                mapping.get('to_concept_url'),
                mapping.get('map_type')
            ]
            csv_writer.writerow(row_data)

    print(f"✅ Success! Wrote {len(mappings_list)} mappings to '{csv_file_path}'")

except FileNotFoundError:
    print(f"❌ Error: The file '{json_file_path}' was not found.")
except Exception as e:
    print(f"❌ An unexpected error occurred: {e}")
```
